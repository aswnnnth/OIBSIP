# **Exploratory Data Analysis (EDA) on Retail Sales Dataset**
**Objective**

The objective of this project is to analyze the retail sales dataset, identify sales trends, compareproduct performance, and generate business insights that support data-driven decision-making.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv ("/content/SampleSuperstore.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

# **Question 1**
What is the overall sales performance of the store?

```python
total_sales = df["Sales"].sum()
print ("Total Sales:", total_sales)
```

# **Observation**
The dataset recorded total sales of **2,296,957.70**(**Approximately
2.3 million**) during the period covered by the dataset.

Business Insight
The strong overall sales indicate healthy business performance. The next step is to identify which regions, product categories, and customer segments contributed the most to this revenue so management can focus resources on the most profitable areas.

# **Question 2**
Which region generated the highest sales?

```python
region_sales = df.groupby ("Region") ["Sales"].sum().sort_values(ascending=False)
region_sales
```

```python
plt.figure(figsize=(8,5))
region_sales.plot(kind="bar")
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=0)
plt.show()
```

# **Observation**
The West region generated the highest total sales ($725,214.6645), followed by the East, Central, and South regions.

Business Insight
The West region is the strongest-performing market in terms of revenue generation. Management should continue investing in this region while studying the strategies that contributed to its success. Similar strategies can be implemented in the South and Central region to improve overall sales performance.

# **Question 3**
Which product category generated the highest sales?

```python
category_sales = df.groupby ("Category")["Sales"].sum().sort_values(ascending=False)
category_sales
```

```python
plt.figure(figsize=(8,5))
category_sales.plot(kind="bar")
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=0)
plt.show()
```

# **Business Insight**
Technology generated the highest total sales ($836,154), making it the best-performing category. Furniture ranked second, while Office Supplies generated the lowest sales among the three categories.

Recommendation
The company should continue investing in Technology products while exploring strategies to improve sales in Office Supplies through promotions, pricing, or product bundling.

# **Question 4**
Which sub-category generated the highest sales?

```python
subcategory_sales = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False)
subcategory_sales
```

```python
plt.figure(figsize=(10,6))
subcategory_sales.plot(kind="bar")
plt.title("Total Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=45)
plt.show()
```

# **Business Insight**
Phones generated the highest total sales ($330,007), followed closely by Chairs. This indicates that customers spend the most on communication devices and offices furniture.

Recommendation
The business should maintain strong inventory levels for Phones and Chairs while promoting lower-performing sub-categories such as Fasteners and Labeles to increase overall sales.

# Question 5
Which customer segment generated the highest sales?

```python
segment_sales = df.groupby ("Segment")["Sales"].sum().sort_values(ascending=False)
segment_sales
```

```python
plt.figure(figsize=(8,5))
segment_sales.plot(kind="bar")
plt.title("Total Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=0)
plt.show()
```

# **Business Insight**
The Consumer segment generated the highest total sales, followed by the Corporate segment. The Home Office segment contributed the last revenue. This suggests that individual consumers are the company's primary source of income and should remain a key focus area for marketing and customer retention strategies.

# **Recommendation**
*Continue targeting the Consumer segment with promotions and loyalty programs.

*Develop tailored offers for Corporate customers to increase repeat purchases.

*Investigate why the Home Office segment has lower sales and introduce    targeted  marketing campaigns or product bundles to improve performance.

# **Final Analysis**
Which product category generated the highest profit?

```python
profit_category = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)
profit_category
```

```python
plt.figure(figsize=(8,5))
profit_category.plot(kind="bar")
plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit ($)")
plt.xticks(rotation=0)
plt.show()
```

# **Business Insight**
Technology generated the highest overall profit, followed by Office Supplies. Although Furniture recorded good sales, it contributed the lowest profit, indicating that this category has lower profit margins or higher operating costs.

# **Recommendation**
*Continue investing in the Technology category through promotions and inventory expansion since it generates the highest profit.

*Maintain strong sales strategies for Office Supplies to sustain profitability.

*Review pricing, discounts, and procurement costs for Furniture to improve its profit margin.

# **Overall Conclusion**
The exploratory data analysis revealed several key business insights:

*Total sales amounted to approximately $2.30 million.

*The West region generated the highest sales, while the South recorded the lowest.

*Technology was the highest-selling and most profitable product category.

*Phones and Chairs were the top-performing sub-categories in terms of sales.

*The Consumer segment contributed the highest revenue, making it the company's most valuable customer group.

*Furniture generated the lowest profit despite relatively high sales, suggesting lower profit margins.
